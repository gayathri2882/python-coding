```python
n=int(input("Enter max no. of candies:"))
k=int(input("Enter min no. of candies:"))
con=1
r=n
while(con==1):
  m=int(input("Enter amount of candies needed:"))
  if(m>n):
    print("invalid")
  else:
    r=r-m
    if(r<=k):
      r=n
    print("Sold=",m)
    print("Remaining=",r)
    con=int(input("press 0 to exit and 1 to continue:"))
```
