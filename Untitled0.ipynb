```python
n = int(input("Enter the number of elements: "))
a = []
for i in range(n):
    m = int(input("Enter the element: "))
    a.append(m)
print("List:", a)
largest = second = float('-inf')
for num in a:
    if num > largest:
        second = largest
        largest = num
    elif num > second and num != largest:
        second = num
print("The largest is", largest)
print("The second largest is", second)
```

```python
r = int(input("Enter number of rats: "))
unit = int(input("Enter the amount of food each rat consumes: "))
n = int(input("Enter number of houses: "))
arr =  __builtins__.list(map(int,input("enter amount of food present in each house: ").split()))

if arr is None or n==0:
    print(-1)
else:
    required = r * unit
    total_food=sum(arr)
    if total_food < required:
      print(0)
    else:
      total=0
      houses=0
      for food in arr:
        total+=food
        houses+=1
        if total>=required:
          break
      print("no of houses",houses)
```
